## 4. Utility functions to reset the face dictionaries and transfer dictionaries by key

```python
def reset_dictionaries(shell):
    faces = Topology.Faces(shell)
    for i, f in enumerate(faces):
        d = Topology.Dictionary(f)
        keys = Dictionary.Keys(d)
        for key in keys:
            if not key == "face_id":
                d = Dictionary.RemoveKey(d, key)
        f = Topology.SetDictionary(f, d)

def transfer_dicts_by_key(topologies, selectors, key):
    dicts = {}
    for t in topologies:
        d = Topology.Dictionary(t)
        value = Dictionary.ValueAtKey(d, key, None)
        if value:
            dicts[str(value)] = t
    
    for s in selectors:
        d = Topology.Dictionary(s)
        value = Dictionary.ValueAtKey(d, key, None)
        if value:
            f = dicts[str(value)]
            f = Topology.SetDictionary(f, d)
```

## 5. Convert to Brep and Import the Gleis 21 ground floor plan

```python
from topologicpy.Topology import Topology
from topologicpy.Cluster import Cluster
from topologicpy.Shell import Shell
from topologicpy.Face import Face

renderer = "vscode"

# 1. Import OBJ file
obj_path = r"Assignment_04.2_Giovanni-Carlo-Volpe_GF_brep.obj"
gleis21_gf_obj = Topology.ByOBJPath(obj_path)
print("Imported objects:", len(gleis21_gf_obj), type(gleis21_gf_obj[0]))

# 2. Cluster and Merge
gleis21_gf_cluster = Cluster.ByTopologies(gleis21_gf_obj)
gleis21_gf_merged = Topology.SelfMerge(gleis21_gf_cluster)  
print("Merged OBJ floorplan:", gleis21_gf_merged)

# 3. Export and import BREP
brep_path = r"Assignment_04.2_Giovanni-Carlo-Volpe_GF.brep"
Topology.ExportToBREP(gleis21_gf_merged, brep_path)
gleis21_gf_brep = Topology.ByBREPPath(brep_path)
print("Imported BREP floorplan:", gleis21_gf_brep)

# 4. Clean up BREP (this might take a few minutes)
triangles = Cluster.Faces(gleis21_gf_brep)

# Create a shell from the triangles
shell = Shell.ByFaces(triangles)

# Extract the external and internal boundaries of the shell
eb = Shell.ExternalBoundary(shell)
ib_list = Shell.InternalBoundaries(shell)

# Create a face from the external boundary and internal boundaries
gleis21_gf_floorplan = Face.ByWires(eb, ib_list)

# Remove collinear edges from the floorplan
gleis21_gf_floorplan = Topology.RemoveCollinearEdges(gleis21_gf_floorplan)
print("Cleaned floorplan Face:", gleis21_gf_floorplan)

# 5. Save the cleaned BREP
clean_brep_path = r"Assignment_04.2_Giovanni-Carlo-Volpe_GF_face.brep"
Topology.ExportToBREP(gleis21_gf_floorplan, clean_brep_path)
print(f"Successfully exported clean face BREP to: {clean_brep_path}")
```

## 14. Spatial Intelligence through Graph Analysis

### Shortest Path (Use navigation graph)

```python
import time

# Define the proportional coordinates of the yellow dots
x_span = xmax - xmin
y_span = ymax - ymin

raw_dots = [
    Vertex.ByCoordinates(xmin + x_span * 0.15, ymax - y_span * 0.08, 0), # 0: Top-Left (path start)
    Vertex.ByCoordinates(xmax - x_span * 0.20, ymax - y_span * 0.10, 0), # 1: Top-Right
    Vertex.ByCoordinates(xmin + x_span * 0.25, ymax - y_span * 0.35, 0), # 2: Mid-Top Left 1
    Vertex.ByCoordinates(xmin + x_span * 0.32, ymax - y_span * 0.35, 0), # 3: Mid-Top Left 2
    Vertex.ByCoordinates(xmin + x_span * 0.05, ymax - y_span * 0.35, 0), # 4: Far Mid-Left protrusion
    Vertex.ByCoordinates(xmin + x_span * 0.28, ymin + y_span * 0.35, 0), # 5: Bottom-Left
    Vertex.ByCoordinates(xmax - x_span * 0.35, ymin + y_span * 0.50, 0), # 6: Mid-Right
    Vertex.ByCoordinates(xmax - x_span * 0.10, ymin + y_span * 0.10, 0)  # 7: Bottom-Right (path end)
]

# Snap dots to the nearest valid graph node to guarantee they are inside boundary
nav_vertices = Graph.Vertices(navigation_graph)
yellow_dots = []
for pt in raw_dots:
    nearest = None
    min_dist = float('inf')
    for nv in nav_vertices:
        dist = Vertex.Distance(pt, nv)
        if dist < min_dist:
            min_dist = dist
            nearest = nv
    yellow_dots.append(nearest)

crg = Graph.CompiledRoutingGraph(navigation_graph, precomputeTurns=False)

start = time.time()
shortest_paths = []
straight_paths = []

# Loop through all dots to find paths between consecutive dots
for i in range(len(yellow_dots) - 1):
    curr_start = yellow_dots[i]
    curr_end = yellow_dots[i+1]
    
    path = Graph.ShortestPath(crg, vertexA=curr_start, vertexB=curr_end)
    if path:
        shortest_paths.append(path)
        st_path = Wire.Straighten(path, host=gleis21_gf_floorplan)
        if st_path:
            straight_paths.append(st_path)

end = time.time()
print("Shortest Path & Straightening Duration:", round(end-start, 2), "seconds")

# Cluster the sequential paths into single geometries
shortest_path = Cluster.ByTopologies(shortest_paths) if shortest_paths else None
straight_path = Cluster.ByTopologies(straight_paths) if straight_paths else None

print("Calculated path segments:", len(shortest_paths))

if straight_path is not None:
    edges = Topology.Edges(straight_path)
    for edge in edges:
        d = Dictionary.ByKeysValues(["width", "color"], [7, "blue"])
        edge = Topology.SetDictionary(edge, d)

if shortest_path is not None:
    edges = Topology.Edges(shortest_path)
    for edge in edges:
        d = Dictionary.ByKeysValues(["width", "color"], [7, "red"])
        edge = Topology.SetDictionary(edge, d)
```

```python
# Color and size the yellow dots for visualization
for pt in yellow_dots:
    d = Dictionary.ByKeysValues(["size", "color"], [15, "orange"])
    Topology.SetDictionary(pt, d)

# Combine the nodes of interest into a cluster for rendering
dots_cluster = Cluster.ByTopologies(yellow_dots)

# Add them to the geometry view alongside the path
if straight_path is not None:
    path_geometry = straight_path
else:
    path_geometry = shortest_path
    
Topology.Show(gleis21_gf_floorplan, shortest_path, straight_path, dots_cluster,
              camera=[0,0,7],
              faceColor=[210,210,250],
              faceOpacity=1,
              edgeColorKey="color",
              edgeWidthKey="width",
              vertexSizeKey='size',
              vertexColorKey='color',
              backgroundColor="black",
              width=800,
              height=600,
              renderer=renderer)
```

* Closeness centrality is a graph metric that quantifies how close a node is to all other nodes by taking the reciprocal of the sum of its shortest path distances to every other node in the network.
* In space syntax, closeness centrality corresponds to global integration, measuring how spatially accessible or topologically shallow a space is within a configuration, thereby indicating its potential for movement flow and encounter density.

### Degree centrality

Bin By Dictionary Key
* Use the community partition number (or colour) to separate the faces of the shell into different bins or categories
* Derive the outer boundary of each face group (perimeter) and make a face out of that perimeter

```python
bins = Topology.BinByDictionaryKey(faces, key="community")
bin_dict = bins[0]
keys = list(bin_dict.keys())
face_groups = []
for key in keys:
    bin_faces = bin_dict[key]
    temp_shell = Shell.ByFaces(bin_faces)
    eb = Shell.ExternalBoundary(temp_shell)
    eb = Wire.RemoveCollinearEdges(eb)
    eb = Face.ByWire(eb)
    face_groups.append(eb)
```
